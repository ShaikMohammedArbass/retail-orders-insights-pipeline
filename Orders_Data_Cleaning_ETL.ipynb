```python
# read data from file and handle null values
import pandas as pd
df = pd.read_csv(r'E:\DataSets\archive (2)\orders.csv',na_values = ['Not Available','unknown'])
df['Ship Mode'].unique()
```

```python
#rename columns names...make them lower case and repalce space with under scores
#df.columns
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df.head(3)
```

```python
#creating new columns called [discount],[sale_price],['profit'] with existing columns list_price and discount_percent
df['discount']=(df['list_price']*df['discount_percent']*.01).round(2)
df['sale_price']=(df['list_price']-df['discount']).round(2)
df['profit']=(df['sale_price']-df['cost_price']).round(2)
df.head(3)
```

```python
#convert order_date from object datatype to (datetime)
df['order_date']=pd.to_datetime(df['order_date'],format="%Y-%m-%d")
```

```python
#drop cost price,list price and discount_percent
df.drop(columns=['cost_price','list_price','discount_percent'],inplace=True,errors='ignore')
```
